```python


import nltk
nltk.download('punkt')  # for sentemce and word tokenization
nltk.download('stopwords')  # for stop word removal
nltk.download('wordnet')    # for Lemmatization
nltk.download('averaged_perceptron_tagger')          # for pos tagging
nltk.download('punkt_tab')    # addtional tokenizer data
nltk.download('omw-1.4') # wordnet language data 
```

```python
text='Natural Language Processing (NLP) is a branch of Artificial Intelligence. It helps computer to understand human language.'
```

```python
from nltk.tokenize import sent_tokenize

sent=sent_tokenize(text)
print(sent)
```

```python
from nltk.tokenize import word_tokenize

words=word_tokenize(text)
print(words)
```

```python
from nltk.corpus import stopwords

stop_words= set (stopwords.words('english'))

filtered_words=[word.lower() for word in words if word not in stop_words]
print(filtered_words)
```

```python
# for word in words:
#     if word not in stop_words:
#         print(word)
```

```python
from nltk.stem import PorterStemmer                # it gives the root word but it didnt give the proper meanigful word
ps=PorterStemmer()

stemmed=[ps.stem(word) for word in filtered_words]
print(stemmed)
```

```python
from nltk.stem import WordNetLemmatizer             # it slower and gives meaningful output it use dictionary word
lem=WordNetLemmatizer()
lemmed=[lem.lemmatize(word) for word in filtered_words]
print(lemmed)
```

```python
words = ['cats','studying','easily']                         # it assume studying as default noun and easily treat as adverb
lemmed=[lem.lemmatize(word) for word in words]
print(lemmed)
```

```python
# pos tag

from nltk.tokenize import word_tokenize
import nltk



nltk.download('averaged_perceptron_tagger_eng')

text='Python is very simple and easy to understand.'

word=word_tokenize(text)

tag=nltk.pos_tag(word)               # it tells the part of speech
print(tag)
```

```python
import spacy                  
```

```python
nlp = spacy.load('en_core_web_sm')         # 
```

```python
text='Rohit Sharma is a captain of Indian Cricket team. He lives in mumbai nad works with BCCI.'
```

```python
# process the text and create a doc object

doc=nlp(text)          
doc
```

```python
for ent in doc.ents:                  # ent => entity ------- it contain specific words(noun)----- (person ,data, time, location , region , city)
    print(ent.text,end=' ')               # GPE  ---- geopolitical entity
    print(ent.label_)
```

```python
from textblob import TextBlob

texts = [
    "Python is easy.",
    "I love Machine Learning.",
    "The weather is nice today."
] 

for text in texts:
    blob = TextBlob(text)
    print("Sentence:", text)
    print("POS Tags:", blob.tags) # POS tagging
    print("Sentiment:", blob.sentiment) # sentiment analysis
    print(blob.words) # tokenization
    print()
```
