```python
import spacy

nlp= spacy.load('en_core_web_sm')
doc = nlp('the smart student solved the problem quickly')

for t in doc:
    print(t.text,t.pos_)
```

```python
# dependancy parsing --- identify how word relate to each other in sentence


doc= nlp('the boy kicked the football')
for t in doc:
    print(t.text,t.dep_,t.head.text)
```

```python
# 1 --- exactly same
# 0 ----- completely unrelated
# -1 ----- opposite direction
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import  cosine_similarity


doc=[
    'I love python programming',
    'I enjoy coding in python',
    'Football is my favourite sport'
]


vect=TfidfVectorizer()
x=vect.fit_transform(doc)
```

```python
sim=cosine_similarity(x)
print(sim)
```

```python
# synsets  -- group of words that shares the meaning

synset= wn.synsets('car')
print(synset)
```

```python
syn = wn.synset('car.n.01')
print(syn.definition())
print(syn.examples())
```

```python
syn = wn.synset('car.n.02')
print(syn.definition())
print(syn.examples())
```

```python
syn = wn.synset('car.n.03')
print(syn.definition())
print(syn.examples())
```

```python
sim=[]

for s in wn.synsets('happy'):
    for lemma in s.lemmas():
        sim.append(lemma.name())


print(set(sim))
        
```

```python
for s in wn.synsets('good'):
    for lemma in s.lemmas():
        if lemma.antonyms():
            print('word :', lemma.name())
            print('ooposite :', lemma.antonyms()[0].name())
```
