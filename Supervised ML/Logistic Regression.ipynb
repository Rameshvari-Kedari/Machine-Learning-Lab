```python
import numpy as np
```

```python

# sample data
X=np.array([[1],[2],[3],[4],[5]])
y=np.array([0,0,0,1,1])
```

```python
# intialized parameter

w=0.0
b=0.0

learning_rate=0.1
epochs=1000
```

```python
#sigmoid

def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
n=len(X)
```

```python
for epoch in range(epochs):
    # linear equation
    z=X.flatten()*w + b

    # prediction 
    y_pred=sigmoid(z)

    #Gradient
    dw=(1/n)*np.sum((y_pred-y)*X.flatten())
    db=(1/n)*np.sum(y_pred-y)

    #update parameter
    w=w-learning_rate*dw
    b=b-learning_rate*db
    
print("weight",w)
print("bias",b)

test=3.5
probability=sigmoid(test*w+b)
print(probability)

if probability>=0.5:
    print("predicted classs:1")

else:
    print("predicted class : 0")
```
